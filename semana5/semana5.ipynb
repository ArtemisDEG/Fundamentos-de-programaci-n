```python
contactos = {
    "Ricardo": "67676767676767",
    "Pedro": "7676767",
    "Luisa": "5656564838290"
}
print(contactos["Pedro"])

contactos["Cris"] = "67674567567"
contactos["Ricardo"] = "12456331134"

print(contactos)

print(contactos.keys())
print(contactos.values())
print(contactos.items())

print("Impresión de llaves")
for n in contactos.keys():
    print(n)

print("\nImpresión de valores")
for v in contactos.values():
    print(v)
print("\n Impresión completa")
for n, t in contactos.items():
    print(f"Nombre: {n}, Telefono: {t}")
```

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "Contacto no encontrado")

nombre = input("Escribe el nombre del contacto que buscas")
telefono = buscar_contacto(contactos, nombre)
print(f"El telefono de {nombre} es: {telefono}")
```

```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0
}
cantidad = int(input("Cuántps votos van a ser?"))
conteo = 1
while conteo <= cantidad:
    voto = input(f"Ingresa el voto {conteo} (A, B, C)")
    if voto.upper() in candidatos:
        print(f"Voto válido para {voto.upper()}")
        candidatos [voto.upper()] +=1
        conteo += 1
for k, v in candidatos.items():
    print(f"Opción {k}, votos {v}")
```
