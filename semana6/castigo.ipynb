#### Ejercicio 10: Suma acumulada con `while`
Pide números mientras sean distintos de `0` (el `0` termina el ciclo y no se suma).

#### Ejercicio 12: Suma de pares del 1 al 20 con `for` y `continue`

#### Ejercicio 15: Matriz y suma de la diagonal principal

````python
matriz=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
suma=0

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        if i == j: 
            suma += matriz[i][j]

print(f"la suma de la diagonal principal es {suma}")
````

#### Ejercicio 20: Diccionario: buscar teléfono de un contacto

````python
directorio = {
    "gussini": "6767070"
}

def buscar_telefono(nombre):
    nombreminuscula = nombre.lower()
    return directorio.get(nombreminuscula, "No se encuentra el pibardo")

contacto = input("Que contacto busca?")
telefono = buscar_telefono(contacto)

if telefono != "No se encuentra el pibardo":
    print(f"El teléfono de {contacto} es: {telefono}")
else:
    print(telefono)

````

#### Ejercicio 21: Diccionario de calificaciones y promedios
