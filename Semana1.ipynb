## Algoritmo para calcular la velocidad

```python
distancia = int(input("Ingresa la distancia en kilometros"))
tiempo = int(input("Ingresa el tiempo en horas"))

print("El tipo de dato de distancia es", type(distancia))
print("El tipo de dato de tiempo es", type(tiempo))

velocidad = distancia / tiempo

print("La velocidad es: ", velocidad)
```
