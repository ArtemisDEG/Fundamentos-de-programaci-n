```python
#Preguntar cuentas calificaciones desea promediar, y promediarlas
def promedio(sumatoria, n):
    return sumatoria / n
n = int(input("¿Cuentas calificaciones quieres promediar?"))
print(f"Se van a pedir {n} calificaciones")
sumatoria = 0
for c in range(1, n+1):
    calif = float(input(f"Dame la calificacion {c}"))
    sumatoria = sumatoria + calif
print(f"El promedio de {n} calificaciones es {promedio(sumatoria, n)}")

```

```python
dimension = int(input("Inserte dimensión"))
matriz = []
for f in range(dimension):
    fila_actual = []
    for c in range(dimension):
        valor = 0
        if f == c:
            valor = 1
        fila_actual.append(valor)
    matriz.append(fila_actual)
print("Impresión de matriz")
for f in range(dimension):
    for c in range(dimension):
        print(matriz[f][c], end="")
    print()
```

```python
lista = []
contador = 1
while contador <= 4:
    valor = int(input("Introduce un número entre -10 y 10"))
    if valor >= -10 and valor <= 10:
        contador += 1
        lista.append(valor)
    else:
        print("Rango incorrecto")
print(lista)
for i in lista:
    if i < 0:
        ass = i * "*"
        print(ass)
    elif i > 0:
        print("Asteriscos positivos")
    else:
        print("|")
```
