```python
matriz = []

for i in range(1, 11):
    fila = []
    for n in range(1, 11):
        fila.append(n * i)
    matriz.append(fila)
def imprimir_tabla(tabla):
    print("\t", end="")
    for n in range(1, 11):
        print(n, end="\t")
    print()
    for i in range(len(tabla)):
        print(i + 1, end="\t")
        
        for numero in tabla[i]:
            print(numero, end="\t")
        print()
imprimir_tabla(matriz)
print("")
renglon = int(input("coord1"))
columna = int(input("coord2"))
resultado = matriz[renglon - 1][columna - 1]
print(f"Coordenada ({renglon}, {columna}) → {resultado}")

```

## EXTRA 1: SUMA DE TODOS LOS ELEMENTOS DE UNA MATRIZ



```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
def suma_matriz(matriz):
    total = 0
    for fila in matriz:
        for numero in fila:
            total = numero + total
    return total

imprimir_matriz(matriz)
print(f"el resultado es: {suma_matriz(matriz)}")
      
```
