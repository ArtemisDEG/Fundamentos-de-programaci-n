## EXTRA 1: SUMA DE TODOS LOS ELEMENTOS DE UNA MATRIZ



```python

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
def suma_matriz(matriz):
    total = 0
    for fila in matriz:
        for numero in fila:
            total = numero + total
    return total

print(f"el resultado es: {suma_matriz(matriz)}")
      
```

## EXTRA 2: MATRIZ CON SUMA POR FILA

```python
matriz=[
    [3,1,4],
    [1,5,9],
    [2,6,5]
]
def mostrar_tabla(matriz):
    for i in matriz:
        for j in i:
            print(j, end=" ")
        print()
mostrar_tabla(matriz)
def suma_fila(fila) :
    sumita = 0                  
    for i in fila:              
        sumita += i
    bro = sumita          
    return bro
for numero_fila in range(len(matriz)):
    bro = suma_fila(matriz[numero_fila])
    print(f"suma de la fila {numero_fila}: {bro}")

    
```

## EXTRA 3: PRODUCTO DE UNA COLUMNA SIN USAR EL OPERADOR *

```python
def multiplicar_columna(matriz, columna):
    producto = 1
    for fila in matriz:
        numero = fila[columna]
        resultado = 0
        for i in range(numero):
            resultado += producto
        producto = resultado
    return producto
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9],
          [2, 4, 6]]
columna = int(input("Ingresa el número de columna (0, 1 o 2): "))
resultado = multiplicar_columna(matriz, columna)
print("El producto de la columna es:", resultado)
```

## EXTRA 4: TABLA DE PITÁGORAS DE TAMAÑO VARIABLE

```python
def generar_tabla(tamano):
    matriz = []
    for renglon in range(1, tamano + 1):
        fila = []
        for columna in range(1, tamano + 1):
            producto = 0
            for i in range(columna):
                producto += renglon
            fila.append(producto)
        matriz.append(fila)
    return matriz
def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero, end="\t")
        print()
tamano = int(input("Ingresa el tamaño de la tabla (2 a 5)"))
if tamano < 2 or tamano > 5:
    print("El tamaño debe estar entre 2 y 5")
else:
    matriz = generar_tabla(tamano)
    imprimir_tabla(matriz)
    renglon = int(input("Ingresa el renglón"))
    columna = int(input("Ingresa la columna"))
    resultado = matriz[renglon - 1][columna - 1]
    print("El producto es", resultado)
```
