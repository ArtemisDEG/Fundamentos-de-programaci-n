## 1. Descripción del reto

Desarrollar un programa en **Python** para cobrar las entradas de los visitantes que desean recorrer el **Museo de Antropología e Historia**, calculando el precio adecuado para cada visitante y aplicando descuentos por tipo de visitante bajo estrictas condiciones lógicas. El programa debe procesar a los visitantes mediante un **ciclo controlado**, aplicar los descuentos con una **tabla de verdad** que garantice un único descuento por boleto y desplegar el **total detallado** de todas las personas ingresadas.

```python
personas = int(input("Cuántos personas nos visitan?"))

for p in range(1, personas + 1):
    edad = int(input(f"Ingresa la edad de la persona {p}"))
    pregun = True
    while pregun:
        if edad < 3:
                print("Pasa gratis")
                pregun = False
        elif edad <= 17:
            estudiante = str(input("Eres estudiante? si/no"))
            if estudiante == "si":
                print(f"Su descuento ha sido considerado, su precio es: ${30 * 0.1}")
                pregun = False
            elif estudiante == "no":
                print(f"Su total es de ${30}")
                pregun = False
            else:
                print("Inválido, vuelva a intentarlo") #Como devolver a la pregunta
        if edad >= 18:
            tipo = str(input("Eres profesor, estudiante o adulto mayor?"))
            while pregun:
                if tipo == "profesor":
                    print(f"Su descuento ha sido considerado, su precio es: ${45 * 0.1}")
                    pregun = False
                elif tipo == "estudiante":
                    print(f"Su descuento ha sido considerado, su precio es: ${45 * 0.1}")
                    pregun = False
                elif tipo == "adulto mayor":
                    print(f"Su descuento ha sido considerado, su precio es: ${45 * 0.12}")
                    pregun = False
                else:
                    print("Inválido, vuelva a intentar") #Como devolver a la pregunta


```
