## 1. Descripción del reto

Desarrollar un programa en **Python** para cobrar las entradas de los visitantes que desean recorrer el **Museo de Antropología e Historia**, calculando el precio adecuado para cada visitante y aplicando descuentos por tipo de visitante bajo estrictas condiciones lógicas. El programa debe procesar a los visitantes mediante un **ciclo controlado**, aplicar los descuentos con una **tabla de verdad** que garantice un único descuento por boleto y desplegar el **total detallado** de todas las personas ingresadas.

```python
preciomenor = 30
precioadulto = 45

descadultomayor = 0.12
descprofe = 0.10
descestu = 0.10

personas = int(input("Cuántas personas nos visitan?"))
total_general = 0

for p in range(1, personas + 1):

    while True:
        edad = int(input(f"Ingresa la edad de la persona {p}: "))

        if edad < 0:
            print("Edad inválida, vuelva a intentarlo")
            continue

        break

    descuento = 0
    tipo = "sin descuento"

    if edad < 3:
        precio_base = 0
        total = 0
        print("Pasa gratis")

    elif edad <= 17:
        precio_base = preciomenor

        estudiante = input("Es estudiante? si/no: ").lower()

        while estudiante != "si" and estudiante != "no":
            print("Respuesta inválida, vuelve a intentarlo")
            estudiante = input("Es estudiante? si/no").lower()

        if estudiante == "si":
            descuento = descestu
            tipo = "estudiante"

        total = precio_base - (precio_base * descuento)

    else:
        precio_base = precioadulto

        tipo_visitante = input(
            "Es profesor, estudiante, adulto mayor u otro? "
        ).lower()

        while (tipo_visitante != "profesor" and
               tipo_visitante != "estudiante" and
               tipo_visitante != "adulto mayor" and
               tipo_visitante != "otro"):

            print("Tipo de visitante inválido, vuelve a intentarlo")
            tipo_visitante = input(
                "Es profesor, estudiante, adulto mayor u otro? "
            ).lower()
        if tipo_visitante == "adulto mayor":
            descuento = descadultomayor
            tipo = "adulto mayor"

        elif tipo_visitante == "profesor":
            descuento = descprofe
            tipo = "profesor"

        elif tipo_visitante == "estudiante":
            descuento = descestu
            tipo = "estudiante"

        total = precio_base - (precio_base * descuento)

    monto_descuento = precio_base * descuento
    total_general = total_general + total

    print(f"Persona {p}")
    print(f"Edad: {edad}")
    print(f"Tipo: {tipo}")
    print(f"Precio base: ${precio_base}")
    print(f"Descuento: ${monto_descuento}")
    print(f"Total a pagar: ${total}")

print(f"Total de todas las personas: ${total_general}")
```
