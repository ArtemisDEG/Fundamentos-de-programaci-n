### Reto semana 2

```python
preciomenor = 30
precioadulto = 45
desacultomayor = 0.12
descprofe = 0.10
descestu = 0.10
personas = int(input("Cuantas personas nos visitan?"))
totalgeneral = 0

for p in range(1, personas + 1):
    while True:
        edad = int(input(f"Ingresa la edad de la persona {p}"))
        if edad < 0:
            print("Edad inválida")
            continue
        break

    descuento = 0
    tipo = "sin descuento"
    if edad <3:
        preciobase = 0
        total = 0
        print("Pasa gratis")
    elif edad <= 17:
        preciobase = preciomenor
        estudiante = input("Eres estudiante= si/no")
        while estudiante != "si" and estudiante != "no":
            print("Respuesta invalida, vuelva a intentarlo")
            estudiante = input("Eres estudiante? si/no")
        if estudiante == "si":
            descuento = descestu
            tipo = "estudiante"
        total = preciobase - (preciobase * descuento)
    else:
        preciobase = precioadulto
        tipovisitante = input("es profesor, estudiante, adulto mayor u otro?")
        while (tipovisitante != "profesor" and tipovisitante != "estudiante" and tipovisitante != "adulto mayor" and tipovisitante != "otro"):
            print("tipo de visitante invalido, vuelve a intentarlo")
            tipovisitante = input("es profesor, estudiante, adulto mayor u otro?")
        if tipovisitante == "adulto mayor":
            descuento = desacultomayor
            tipo = "adulto mayor"
        elif tipovisitante == "profesor":
            descuento = descprofe
            tipo = "profesor"
        elif tipovisitante == "estudiante":
            descuento = descestu
            tipo = "estudiante"
        total = preciobase - (preciobase * descuento)

    montodescuento = preciobase * descuento
    totalgeneral = totalgeneral + total

    print("Tabla")
    print(f"Persona {p}")
    print(f"Edad: {edad}")
    print(f"Tipo: {tipo}")
    print(f"Precio base: $ {montodescuento}")
    print(f"Total a pagar: ${total}")
print(f"Total de todas las personas ${totalgeneral}")
                
```

## EXTRA 2

```python
visitante = int(input("Cuantas personas ingresaran?"))

mayor = 0
edades = 0

for p in range(1, visitante + 1):
    edad = int(input(f"Ingresa la edad de la persona {p}"))
    if edad > 0:
        edades += edad
    if edad >= 18:
        mayor += 1
promedio = edades/visitante
print(f"El promedio de edad de los visitantes es de {promedio}")
print(f"El numero de mayores de edad es de {mayor}")
```
