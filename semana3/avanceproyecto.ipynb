## Avance proyecto

Empresa/organización: Qrobús
Problematica: Viajes con falta de saldo en tarjeta de prepagfo
Descripción: 560 mil personas usan el Qrobús, convirtiendo a este sistema de transporte público en uno de los más modernos y grandes del país, sin embargo, al usar tarjetas de prepago, la falta de saldo en esta es un problema común en muchos de nosotros, y una desventaja de esto es no poder realizar viajes debido a esta falta de saldo, por ello, se propone implementar un sistema de adeudo, dónde el pasajero tenga un máximo de 2 viajes consecutivos sin saldo, la deuda se agregará al historial de la tarjeta, dónde se descontará de la siguiente recarga, esto permitirá al usuario llegar a su destino, a alguna tienda de conveniencia o estación dónde pueda recargar la tarjeta. Después de los 2 viajes sin saldo, la tarjeta se bloqueará y se desbloqueará al pagar el adeudo.

Si viajes es menor a 2 entonces

    aceptar viaje

    print "Saldo insuficiente, se realizó el viaje"

    denegar viaje
